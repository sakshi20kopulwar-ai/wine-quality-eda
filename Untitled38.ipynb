```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
sns.set_theme(style="whitegrid")
```

```python
PATH = "Wine Quality dataset.csv"

def load(path):
    try:
        return pd.read_csv(path)
    except (UnicodeDecodeError, pd.errors.ParserError):
        return pd.read_excel(path, engine="openpyxl")   # file is actually .xlsx

df = load(PATH)
print("Shape:", df.shape)
print(df.head())
```

```python
print(df.tail())
```

```python
df.info()
print(df.describe().T)
print("Missing values:\n", df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.columns = (df.columns.str.strip().str.lower()
                .str.replace(r"\s+", "_", regex=True))

# 3b. strip whitespace in any text columns
for c in df.select_dtypes("object"):
    df[c] = df[c].str.strip()
```

```python
df = df.drop_duplicates().reset_index(drop=True)
df
```

```python
num_cols = df.select_dtypes(include="number").columns.tolist()
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
def iqr_bounds(s, k=1.5):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr

outlier_counts = {c: int(((df[c] < iqr_bounds(df[c])[0]) |
                          (df[c] > iqr_bounds(df[c])[1])).sum())
                  for c in num_cols}
print("Outliers per column:", outlier_counts)
```

```python
df_capped = df.copy()
for c in num_cols:
    if c == "quality":
        continue
    lo, hi = iqr_bounds(df_capped[c])
    df_capped[c] = df_capped[c].clip(lo, hi)
```

```python
mask = pd.Series(True, index=df.index)
for c in num_cols:
    if c == "quality":
        continue
    lo, hi = iqr_bounds(df[c], k=3)          # k=3 -> only extreme values
    mask &= df[c].between(lo, hi)
df_no_out = df[mask]
```

```python
print(f"Rows kept after removing extreme outliers: {len(df_no_out)} / {len(df)}")
```

```python
df["quality_label"] = pd.cut(df["quality"], bins=[0, 5, 6, 10],
                             labels=["low", "medium", "high"])
df["sulfur_ratio"] = df["free_sulfur_dioxide"] / df["total_sulfur_dioxide"]
```

```python
# 5. FILTERING
# ------------------------------------------------------------------
high_q      = df[df["quality"] >= 7]
strong      = df[(df["alcohol"] > 12) & (df["quality"] >= 7)]
either      = df[(df["ph"] < 3.0) | (df["ph"] > 3.6)]
in_range    = df[df["alcohol"].between(10, 12)]
in_list     = df[df["quality"].isin([3, 4])]
by_query    = df.query("alcohol > 11 and residual_sugar < 5 and quality >= 6")
top_alcohol = df[df["alcohol"] > df["alcohol"].quantile(0.9)]   # top 10%
print(len(high_q), len(strong), len(by_query))
```

```python
# 6. SORTING / RANKING
# ------------------------------------------------------------------
print(df.sort_values("alcohol", ascending=False).head(10))
print(df.sort_values(["quality", "alcohol"], ascending=[False, False]).head(10))
print(df.nlargest(5, "residual_sugar"))
print(df.nsmallest(5, "ph"))
df["alcohol_rank"] = df["alcohol"].rank(ascending=False, method="min")
```

```python
# 7. GROUPING / AGGREGATION
# ------------------------------------------------------------------
print(df.groupby("quality")[num_cols].mean().round(3))
print(df.groupby("quality_label", observed=True).agg(
    count=("quality", "size"),
    avg_alcohol=("alcohol", "mean"),
    avg_sugar=("residual_sugar", "mean"),
    max_ph=("ph", "max")))

df["alcohol_band"] = pd.qcut(df["alcohol"], 4, labels=["Q1", "Q2", "Q3", "Q4"])
print(pd.crosstab(df["alcohol_band"], df["quality_label"], normalize="index").round(3))
print(df.pivot_table(index="alcohol_band", columns="quality_label",
                     values="residual_sugar", aggfunc="mean", observed=True))
```

```python
# 8. STATISTICAL TESTS (optional)
# ------------------------------------------------------------------
from scipy import stats
hi = df.loc[df["quality"] >= 7, "alcohol"]
lo = df.loc[df["quality"] <= 5, "alcohol"]
t, p = stats.ttest_ind(hi, lo, equal_var=False)
print(f"Alcohol, high vs low quality: t={t:.2f}, p={p:.3g}")
```
